##About This Project
In this project, I analyzed global temperature data from 1750 to 2015.  
I used Python to clean the data, explore patterns , and create visualizations to understand how Earth's temperature has changed over time.

```python
import pandas as pd

df=pd.read_csv("GlobalTemperatures.csv")
df.head()
```

```python
print("Rows and Columns: ",df.shape)
```

```python
print(df.columns.tolist())
```

```python
print(df.isnull().sum())
```

```python
df.dropna(inplace=True)
df["dt"]=pd.to_datetime(df["dt"])
df["Year"]=df["dt"].dt.year
print("Data is clean and ready!")
```

```python
import matplotlib.pyplot as mpl
yearly_avg = df.groupby("Year")["LandAverageTemperature"].mean()

mpl.figure(figsize=(14,5))
mpl.plot(yearly_avg.index,yearly_avg.values, color="tomato")
mpl.title("Global Average Temperature Over Time")
mpl.xlabel("Year")
mpl.ylabel("Temperature(˚C)")
mpl.grid(True)
mpl.show()
```

## Finding 1
Global temperature has been rising steadily since the 1900s.  
The increase is especially sharp after 1980, which suggests the impact of industrialization and carbon emissions.

```python
top_years=yearly_avg.nlargest(10)
mpl.figure(figsize=(10,5))
mpl.bar(top_years.index, top_years.values, color="orange")
mpl.title("Top 10 Hottest Years Ever")
mpl.xlabel("Year")
mpl.ylabel("temperature(˚C)")
mpl.show()
```

## Finding 2
All of the top 10 hottest years recorded were between 1997 and 2015.  
This confirms that recent decades are significantly warmer than any previous period in history.

```python
mpl.figure(figsize=(10,5))
mpl.hist(df["LandAverageTemperature"],bins=40, color="steelblue")
mpl.title("Temperature Distribution")
mpl.xlabel("Temperature(˚C)")
mpl.ylabel("Frequency")
mpl.show()
```

## Finding 3
The temperature distribution shows 3 distinct peaks, ranging from 0˚C to 16˚C.  This pattern exists because the dataset includes all seasons - winter months pull temperatures low (around 2-4˚C) while summer months push them high (around 13-15˚C). The most frequent temperatures appear near 14˚C, representing warmer periods.

##Conclusion
This analysis clearly shows that global temperatures are rising over time.  
The data strongly supports the reality of climate change.
